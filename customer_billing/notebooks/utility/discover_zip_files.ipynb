```python
# Discover ZIP files in ADLS
dbutils.widgets.text("input_folder", "")
input_folder = dbutils.widgets.get("input_folder")
if not input_folder:
    raise ValueError("Input folder parameter is empty")
input_folder = input_folder + "/"
print("Input folder received:", input_folder)
zip_files = [
    {"file_name": file.name, "zip_path": file.path}
    for file in dbutils.fs.ls(input_folder)
    if file.name.lower().endswith(".zip")
]
if not zip_files:
    raise ValueError(f"No ZIP files found in {input_folder}")
print(f"Total ZIP files found: {len(zip_files)}")
for item in zip_files:
    print(item)
dbutils.jobs.taskValues.set(key="zip_files", value=zip_files)
```
